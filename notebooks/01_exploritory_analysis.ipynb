```python
import pandas as pd
import glob

# Load ATP match data from 2020 through 2024
files = sorted(glob.glob('../data/tennis_atp/atp_matches_202*.csv'))
df_list = [pd.read_csv(f) for f in files]
df = pd.concat(df_list, ignore_index=True)

# Filter out matches with missing rankings
valid_matches = df.dropna(subset=['winner_rank', 'loser_rank']).copy()

# Calculate how often the higher-ranked player wins (smaller numerical rank = better position)
higher_rank_won = (valid_matches['winner_rank'] < valid_matches['loser_rank']).sum()
total_matches = len(valid_matches)
baseline_acc = higher_rank_won / total_matches

print(f"Total Matches Analyzed (2020-2024): {total_matches}")
print(f"Higher-Ranked Player Win Rate (Baseline Benchmark): {baseline_acc:.2%}")
```
